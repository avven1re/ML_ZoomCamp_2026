```python
import numpy as np
```

### Creating arrays

```python
np.zeros(5)
```

```python
np.ones(10)
```

```python
np.full(10, 2.5) # fill the number 2.5 in the size of 10 array
```

```python
a = np.array([1, 2, 3, 5, 7, 12])
a
```

```python
a[2] = 10
a
```

```python
np.arange(3, 10) # create 3-9 array, 10 is exclusive, similar to python func. range()
```

```python
np.linspace(0, 10, 11) # creating a array filled with the number in 0 to 10 (inclusive) with size of 11
```

### Multi-dimensional arrays

```python
np.zeros((5, 2)) # 5 rows and 2 colmns filled with zeros
```

```python
n = np.array([
    [1, 2, 3], #0
    [4, 5, 6], #1
    [7, 8, 9], #2
])  #0 #1 #2
n
```

```python
n[1][2] # not recommended
```

```python
n[1, 2] # recommended
```

```python
n[2] = [1, 1, 1]
n
```

```python
n[:, 2]
```

### Randomly generated arrays

```python
np.random.seed(2)
np.random.rand(5, 2)
```

```python
np.random.seed(2)
np.random.randn(5, 2) # sample from standard normal dist.
np.random.randint(low = 0, high = 100, size = (5, 2))
```

### Element-wise operations

```python
a = np.arange(5)
a
```

```python
a + 1 # in usual python list, this operation cannot be executed
```

```python
a * 2
```

```python
b = (10 + (a * 2)) ** 2 / 100
```

```python
a * b
```
