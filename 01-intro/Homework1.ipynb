## Q6. Median value of horsepower

1. Find the median value of the `horsepower` column in the dataset
2. Next, calculate the most frequent value of the same `horsepower` column
3. Use the `fillna` method to fill the missing values in the `horsepower` column with the most frequent value from the previous step
4. Now, calculate the median value of `horsepower` once again

Has it changed?

```python
# Step 1
df['horsepower'].median()
```

the median of `horsepower` is 254

```python
# Step 2
df['horsepower'].mode()
```

the most frequent value of `horsepower` is 252

```python
# Step 3
mode_val = float(df['horsepower'].mode().iloc[0])
print(mode_val)
df['horsepower'] = df['horsepower'].fillna(mode_val)
```

```python
# step 4
df['horsepower'].median()
```

**Ans: Yes, it decreased.**

**We can see that the median of `horsepower` decreased from 254 to 252, after we filled the missing values with the most frequent value of `horsepower`: 252**

## Q7. Sum of weights
1. Select all the cars from Asia
2. Select only columns `vehicle_weight` and `model_year`
3. Select the first 7 values
4. Get the underlying NumPy array. Let's call it `X`.
5. Compute matrix-matrix multiplication between the transpose of `X` and `X`. To get the transpose, use `X.T`. Let's call the result `XTX`.
6. Invert `XTX`.
7. Create an array `y` with values [1100, 1300, 800, 900, 1000, 1100, 1200].
8. Multiply the inverse of `XTX` with the transpose of `X`, and then multiply the result by `y`. Call the result `w`.
9. What's the sum of all the elements of the result?

```python
# step 1 - 3
df_asia = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
df_asia
```

```python
# step 4
X = np.array(df_asia)
X
```

```python
# step 5 - 6
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
print(XTX_inv)
```

```python
# step 7
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
# step 8 - 9
temp = XTX_inv.dot(X.T)
w = temp.dot(y)
w.sum()
```

**Ans: the sum of w is 0.369**
